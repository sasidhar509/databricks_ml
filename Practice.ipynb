```python
from pyspark.sql.functions import col
```

```python
df=df.filter(col("Age") > 18)
df.display()
```

```python
df2 = df.filter(col("Total Amount") > 100)          # Narrow
df3 = df2.select("Customer ID", "Total Amount") # Narrow
df4 = df3.groupBy("Customer ID").sum("Total Amount") # Wide
df4.display()   
```

```python
cust = spark.createDataFrame([("C1","Alice"),("C2","Bob")], ["CustomerID","Name"])
trans = spark.createDataFrame([("T1","C1",100),("T2","C2",200),("T3","C3",50)],
                              ["TransID","CustomerID","Amount"])
cust.display()
trans.display()
```

```python
inner_join=cust.join(trans,on="CustomerID",how="inner")

inner_join.show()
```

```python
left_join=cust.join(trans,on="CustomerID",how="left")
left_join.show()
```

```python
right_join=cust.join(trans,on="CustomerID",how="right")
right_join.show()
```

```python
full_join=cust.join(trans,on="CustomerID",how="full")
full_join.show()
```

```python
cross = cust.crossJoin(trans)  # or .join(trans, how='cross')
cross.show()
```

```python
from pyspark.sql.functions import broadcast
df_join = broadcast(cust).join(trans, "CustomerID")
df_join.show()
```

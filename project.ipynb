```python
from pyspark.sql import functions as F

top_customer = (
    df.filter(F.year("Date") == 2024)
      .groupBy("Customer ID")
      .agg(F.sum("Total Amount").alias("sum_amount"))
      .orderBy(F.desc("sum_amount"))
      .limit(2)
)

display(top_customer.select("Customer ID"))
```
